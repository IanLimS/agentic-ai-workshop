## 2. 환경 설정

```python
import os
os.environ['ADP_BASE_URL'] = "https://api.lkeap.tencentcloud.com/v1"
os.environ['ADP_API_KEY'] = "YOUR_API_KEY"  # 실제 키로 변경 필요
os.environ['ADP_MODEL'] = "deepseek-r1"
```

## 3. API 예제

```python
import requests

headers = {
    "Authorization": f"Bearer {os.getenv('ADP_API_KEY')}",
    "Content-Type": "application/json"
}

data = {
    "model": os.getenv("ADP_MODEL"),
    "messages": [
        {"role": "system", "content": "지식베이스만 사용해 답변하세요"},
        {"role": "user", "content": "반품 정책을 알려주세요"}
    ]
}

response = requests.post(
    f"{os.getenv('ADP_BASE_URL')}/chat/completions",
    headers=headers,
    json=data
)
print(response.json())
```
